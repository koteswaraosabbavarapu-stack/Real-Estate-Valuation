# Real Estate Valuation & Automated Appraisal Engine
## Exploratory Data Analysis (EDA)

This notebook explores the Ames Housing dataset to understand key property valuation drivers, missing value distributions, correlation patterns, and target variable skewness.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style='whitegrid')
plt.rcParams.update({'font.size': 11})

train_df = pd.read_csv('../data/raw/train.csv')
print(f'Training Data Shape: {train_df.shape}')
train_df.head()
```

### 1. Target Variable: SalePrice Analysis
Let's inspect the distribution of `SalePrice` and verify whether a `log1p` transformation normalizes its right-skew.

```python
sp = train_df['SalePrice']
print('SalePrice Summary Statistics:')
print(sp.describe())
print(f'Original Skewness: {sp.skew():.4f}')
print(f'Log-Transformed Skewness: {np.log1p(sp).skew():.4f}')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(sp, kde=True, ax=axes[0], color='#2b5c8f')
axes[0].set_title('Original SalePrice Distribution (Right-Skewed)')

sns.histplot(np.log1p(sp), kde=True, ax=axes[1], color='#2e8b57')
axes[1].set_title('Log-Transformed log1p(SalePrice) (Normal Distribution)')
plt.show()
```

### 2. Key Valuation Drivers
Let's visualize the relationship between overall quality, living area, year built, and house prices.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.boxplot(x='OverallQual', y='SalePrice', data=train_df, ax=axes[0], palette='viridis')
axes[0].set_title('Overall Quality Rating (1-10) vs SalePrice')

sns.regplot(x='GrLivArea', y='SalePrice', data=train_df, ax=axes[1],
            scatter_kws={'alpha':0.4, 'color':'#1f77b4'}, line_kws={'color':'#d62728'})
axes[1].set_title('Above Grade Living Area (GrLivArea) vs SalePrice')
plt.show()
```

### 3. Missing Value Analysis

```python
missing = train_df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
missing_pct = (missing / len(train_df)) * 100
missing_summary = pd.DataFrame({'Missing_Count': missing, 'Percentage': missing_pct})
missing_summary.head(15)
```
